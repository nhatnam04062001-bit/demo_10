```python
import pandas as pd

# Series - mảng 1 chiều - 1 cột dữ liệu
s1 = pd.Series([10,20,30])
print(s1)
print("index :", s1.index)
print("values :", s1.values)
```

```python
# Dataframe - mảng 2 chiều - nhiều cột dữ liệu
# tạo df từ dictionary
data = {
    "Product" : ["Laptop", "Phone", "Tablet"],
    "Price": [1000, 800, 500]
}
df = pd.DataFrame(data)
print(df)
```

```python
# Tạo dataframe từ list of dictionaries
datalist = [
    {"product" : "Laptop", "Price" : 1200},
    {"product" : "Phone", "Price" : 1000},
]
df = pd.DataFrame(datalist)
print(df)
```

```python
# đọc dữ liệu từ file csv
df = pd.read_csv("Sale.csv")
print(df)
```

```python
# khám phá dữ liệu
# hiện thị 3 dòng đầu tiên
print(df.head(3))
```

```python
# hiển thị 3 dòng cuối cùng
print(df.tail(3))
```

```python
# kích thước của df
print(df.shape)
```

```python
#xem thông tin của df
print(df.info())
```

```python
# thông tin cơ bản của df
print(df.describe())
```

```python
# thêm 1 cột vào df
df["Subtotal"] = df["Price"] * df["Quantity"]
print(df)
```

```python
# kiểm tra missing value
# kiểm tra ô nào bị thiếu
print(df.isnull())
```

```python
# đếm số lượng missing của từng cột
print(df.isnull().sum())
```

```python
# Loại bỏ những dòng có missing
dfDrop = df.dropna()
print(df)
print(dfDrop)
```

```python
# Điền dữ liệu còn thiếu
dfFill = df.copy()

# Điền giá còn thiếu bằng giá nhỏ nhất trong cột Price
dfFill["Price"] = dfFill["Price"].fillna(dfFill["Price"].min())

# Điền thành phố còn thiếu bằng "Unknown"
dfFill["City"] = dfFill["City"].fillna("Unknown")

print(dfFill)
```

```python
# kiểm tra duplicate
print(df.duplicated)
```

```python
# Xóa duplicate
dfClean = df.drop_duplicates()
print(df)
print(dfClean)
```
